```python
def max_increasing(arr: list):
  best, curr = [], []
  for x in arr:
    if not curr: curr = [x]
    if x > curr[-1]: curr.append(x)
    if len(best)<len(curr): best = curr
    else: curr = [x]
  return best

def fast_max_increasing(arr: list):
  if not arr: return []
  best, curr = (0, 1), [0, 1]
  dist = lambda x: x[1]-x[0]
  for i in range(1, len(arr)):
    if arr[i]>arr[curr[1]-1]: curr[1]+=1
    if dist(best)<dist(curr): best = curr.copy()
    else: curr = [i, i+1]
  return arr[best[0]:best[1]]

def test_f(f):
  assert f([]) == []
  assert f([1, 2]) == [1, 2]
  assert f([4, 2, 3, 5, 5, 6, 4, 5]) == [2, 3, 5]

test_f(max_increasing)
test_f(fast_max_increasing)
```
